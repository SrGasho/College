```python
import sys, pathlib
sys.path.insert(0, str(pathlib.Path.cwd().parent))
import pandas as pd
import matplotlib.pyplot as plt
from src import config as C

data_cfg = C.load_data_config('../configs/data.yaml')
interim = C.data_path(data_cfg, 'interim')
frames = {cid: pd.read_parquet(interim / f'frames_{cid}.parquet') for cid in C.capture_ids(data_cfg)}
episodes = pd.read_parquet(interim / 'episodes_exploratory.parquet')
episodes.groupby('capture_id').size()
```

```python
# Preguntas 2 y 11: ¿ráfaga continua o varias ráfagas?
for cid, df in frames.items():
    t = df.loc[df['is_target'], 'ts']
    fig, ax = plt.subplots(figsize=(8, 2))
    ax.hist(t, bins=200)
    ax.set_title(f'Tramas objetivo en el tiempo: {cid}')
    ax.set_xlabel('ts (s)')
    plt.show()
```

```python
# Preguntas 7, 8 y 14: subtipos, códigos de motivo y desconexiones no objetivo
for cid, df in frames.items():
    disc = df[(df['fc_type'] == 0) & df['fc_subtype'].isin([10, 12])]
    print(cid)
    print(pd.crosstab(disc['reason_code'], [disc['fc_subtype'], disc['is_target']]))
```
